```python
def naive_add(x, y):
  assert len(x.shape) == 2
  assert x.shape == y.shape
  x = x.copy()
  for i in range(x.shape[0]):
    for j in range(x.shape[1]):
      x[i, j] += y[i, j]
  return x

def naive_relu(x):
  assert len(x.shape) == 2
  x = x.copy()
  for i in range(x.shape[0]):
    for j in range(x.shape[1]):
      x[i, j] = max(x[i, j], 0.0)
  return x

t0 = time.time()
for _ in range(1000):
  z = naive_add(x, y)
  z = naive_relu(z)
print("Took: {0: .2f} s".format(time.time() - t0))
```

```python
def naive_vector_product(x, y):
  assert len(x.shape) == 1
  assert len(y.shape) == 1
  assert x.shape[0] == y.shape[0]
  z = 0.0
  for i in range(x.shape[0]):
    z += x[i] * y[i]
  return z
```

```python
def native_matrix_vector_product(x, y):
  assert len(x.shape) == 2
  assert len(y.shape) == 1
  assert x.shape[1] == y.shape[0]
  z = np.zeros(x.shape[0])
  for i in range(x.shape[0]):
    for j in range(x.shape[1]):
      z[i] += x[i, j] * y[j]
  return z
```

```python
def native_matrix_vector_product(x, y):
  z = np.zeros(x.shape[0])
  for i in range(x.shape[0]):
    z[i] = naive_vector_product(x[i, :], y)
  return z
```

```python
def naive_matrix_product(x, y):
  assert len(x.shape) == 2
  assert len(y.shape) == 2
  assert x.shape[1] == y.shape[0]
  z = np.zeros((x.shape[0], y.shape[1]))
  for i in range(x.shape[0]):
    for j in range(y.shape[1]):
      row_x = x[i, :]
      column_y = y[:, j]
      z[i, j] = naive_vector_product(row_x, column_y)
  return z
```

### 行列の shape と、切り出したときの「要素数」を検証するコード

3行5列の行列 `y` を作成して、以下の2つを比較・検証します：
1. **横方向（行）**を取り出した時の要素数（＝ 列数 5 と一致するはず）
2. **縦方向（列）**を取り出した時の要素数（＝ 行数 3 と一致するはず）

```python
import numpy as np

# 3行5列の行列 Y を作成します (1から15までの数字)
Y = np.arange(1, 16).reshape(3, 5)

print("元の行列 Y:")
print(Y)
print(f"Y の形状 (shape): {Y.shape} -> ({Y.shape[0]}行, {Y.shape[1]}列)\n")

# -----------------------------------------
# 1. 横方向 (行) を取り出す
# -----------------------------------------
row_0 = Y[0, :] # 0行目のすべての列を取り出す
print("--- 1. 横方向 (0行目) を取り出す ---")
print(f"取り出した行ベクトル: {row_0}")
print(f"このベクトルの要素数 (len): {len(row_0)}")
print(f"元の行列の『列数 (Y.shape[1])』: {Y.shape[1]}")
print(f"一致しているか?: {len(row_0) == Y.shape[1]}\n")

# -----------------------------------------
# 2. 縦方向 (列) を取り出す
# -----------------------------------------
col_0 = Y[:, 0] # すべての行から 0列目を取り出す
print("--- 2. 縦方向 (0列目) を取り出す ---")
print(f"取り出した列ベクトル: {col_0}")
print(f"このベクトルの要素数 (len): {len(col_0)}")
print(f"元の行列の『行数 (Y.shape[0])』: {Y.shape[0]}")
print(f"一致しているか?: {len(col_0) == Y.shape[0]}")
```
